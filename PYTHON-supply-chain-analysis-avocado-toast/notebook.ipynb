# Co kryje się w avocado toast: analiza łańcucha dostaw

![](avocado_wallpaper.jpeg)

Jesteś w Londynie i szykujesz pyszny avocado toast — danie, które od lat 2010 mocno zyskało na popularności w menu śniadaniowym. Ten prosty przepis wymaga tylko pięciu składników: dojrzałego awokado, połówki cytryny, porządnej szczypty soli w płatkach, dwóch kromek chleba na zakwasie i solidnej porcji oliwy extra vergine. Większość z tych składników to dziś sklepowe pewniaki — i, jak zobaczysz w tym projekcie, to wcale nie jest takie oczywiste!

W tym projekcie przeanalizujesz łańcuch dostaw trzech składników używanych w avocado toast, korzystając z bazy Open Food Facts. Ta baza zawiera obszerne, otwarte informacje o różnych produktach spożywczych, w tym o ich pochodzeniu. Dzięki tej analizie lepiej zrozumiesz złożony łańcuch dostaw stojący za przygotowaniem jednego dania.

W folderze z danymi znajdziesz trzy pary plików:
- Plik CSV dla każdego składnika, np. `avocado.csv`, z danymi o każdym produkcie i krajach pochodzenia.
- Plik TXT dla każdego składnika, np. `relevant_avocado_categories`, zawierający tylko interesujące nas tagi kategorii dla danego produktu.

Kilka kluczowych uwag o tych plikach:
- Część wierszy w każdym z trzech plików CSV nie zawiera danych istotnych dla twojego badania. W każdym zbiorze musisz odfiltrować nieistotne wiersze na podstawie wartości w kolumnie `categories_tags`. Przykłady kategorii to owoce, warzywa i oleje na bazie owoców. Przefiltruj DataFrame tak, by zostawić tylko wiersze, w których `categories_tags` zawiera któryś z tagów z listy właściwych kategorii dla danego składnika.
- W kolumnie `categories_tags` zwykle jest wiele tagów kategorii w jednym wierszu.
W każdym pliku CSV jest kolumna `origins_tags`, która zawiera łańcuchy opisujące kraj pochodzenia danego produktu.

Po ukończeniu projektu będziesz mieć listę składników i ich krajów pochodzenia, a także solidną bazę do dalszych analiz — na przykład ile średnio czasu spędzają te składniki w transporcie morskim.

[Open Food Facts database](https://world.openfoodfacts.org/)

```python
import pandas as pd

# table headers
head_tabel = ['code', 'lc', 'product_name_en', 'quantity', 'serving_size', 'packaging_tags', 'brands', 'brands_tags', 'categories_tags', 'labels_tags', 'countries', 'countries_tags', 'origins', 'origins_tags']

# load csv with categories in txt
avocado = pd.read_csv("data/avocado.csv", sep="\t")
avocado_categories = pd.read_csv("data/relevant_avocado_categories.txt", header=None)[0].tolist()
avocado2 = avocado[head_tabel]
avocado3 = avocado2[avocado2['categories_tags'].str.contains("|".join(avocado_categories), na=False)]

olive_oil = pd.read_csv("data/olive_oil.csv", sep="\t")
olive_oil_categories = pd.read_csv("data/relevant_olive_oil_categories.txt", header=None)[0].tolist()
olive_oil2 = olive_oil[head_tabel]
olive_oil3 = olive_oil2[olive_oil2['categories_tags'].str.contains("|".join(olive_oil_categories), na=False)]

sourdough = pd.read_csv("data/sourdough.csv", sep="\t")
sourdough_categories = pd.read_csv("data/relevant_sourdough_categories.txt", header=None)[0].tolist()
sourdough2 = sourdough[head_tabel]
sourdough3 = sourdough2[sourdough2['categories_tags'].str.contains("|".join(sourdough_categories), na=False)]

# data filtering
avocado_uk = avocado3[avocado3['countries'] == 'United Kingdom']
top_avocado_origin = avocado_uk['origins_tags'].value_counts().sort_values(ascending=False).index[0]

olive_oil_uk =  olive_oil3[olive_oil3['countries'] == 'United Kingdom']
top_olive_oil_origin = olive_oil_uk['origins_tags'].value_counts().sort_values(ascending=False).index[0]

sourdough_uk =  sourdough3[sourdough3['countries'] == 'United Kingdom']
top_sourdough_origin = sourdough_uk['origins_tags'].value_counts().sort_values(ascending=False).index[0]

# function for removing extra characters
def zamiana (zmienna):
    return zmienna.replace("en:", "").replace("-", " ")

# results
top_avocado_origin = zamiana(top_avocado_origin)
top_olive_oil_origin = zamiana(top_olive_oil_origin)
top_sourdough_origin = zamiana(top_sourdough_origin)


```
